#### Definiciones

```python
# Circuito
'''
            |------ZA-------|
            |               |
    1-------+--R----2---R---3
                    |       |
                   ZB       R
                    |       |
    4---------------+--------
'''
#Simbolos
ZA = sp.symbols('Za', real = True, positive = True)
ZB = sp.symbols('Zb', real = True, positive = True)
R = sp.symbols('R', real = True, positive = True)

#MAI
y_mai = sp.Matrix( [
    [1/ZA + 1/R, -1/R, -1/ZA, 0],
    [-1/R, 2/R + 1/ZB, -1/R, -1/ZB],
    [-1/ZA, -1/R, 2/R + 1/ZA, -1/R],
    [0, -1/ZB, -1/R, 1/ZB+1/R] 
    ])
```

#### Resolución

```python
z_in = calc_MAI_impedance_ij(y_mai, 1-1, 4-1)
z_in_fx = z_in.subs(ZA*ZB, R**2)
print(f"La impedancia de entrada queda: ")
z_in_fx
```

#### _yt con sympy_


```python
#Script
#Simbolos
ZA = sp.symbols('Za', real = True, positive = True)
ZB = sp.symbols('Zb', real = True, positive = True)
R = sp.symbols('R', real = True, positive = True)
#Armo Y1 e Y2
y1 = sp.Matrix ([
    [(R+ZB)/(R*R+R*2*R*ZB), -(ZB)/(R*R+R*2*R*ZB)],
    [-(ZB)/(R*R+R*2*R*ZB), (R+ZB)/(R*R+R*2*R*ZB)]
    ])
y2 = sp.Matrix ([
    [1/ZA, -1/ZA],
    [-1/ZA, (ZA+R)/(ZA*R)]
    ])
#Multiplico ambas matrices
yt = y1*y2
#Reemplazo R, ZA y ZB por 1
yt_fx = yt.subs({R:1, ZA:1, ZB:1})
yt_fx
```
