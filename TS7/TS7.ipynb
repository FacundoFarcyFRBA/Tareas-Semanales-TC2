#### Definiciones

```python
# Circuito
'''
            |------ZA-------|
            |               |
    1-------+--R----2---R---3
                    |       |
                   ZB       R
                    |       |
    4---------------+--------
'''
#Simbolos
ZA = sp.symbols('Za', real = True, positive = True)
ZB = sp.symbols('Zb', real = True, positive = True)
R = sp.symbols('R', real = True, positive = True)

#MAI
y_mai = sp.Matrix( [
    [1/ZA + 1/R, -1/R, -1/ZA, 0],
    [-1/R, 2/R + 1/ZB, -1/R, -1/ZB],
    [-1/ZA, -1/R, 2/R + 1/ZA, -1/R],
    [0, -1/ZB, -1/R, 1/ZB+1/R] 
    ])
```

#### Resolución

```python
z_in = calc_MAI_impedance_ij(y_mai, 1-1, 4-1)
z_in_fx = z_in.subs(ZA*ZB, R**2)
print(f"La impedancia de entrada queda: ")
z_in_fx
```

### <u>Comprobación con Simpy</u>

```python
#Script
#Simbolos
C = sp.symbols('C', real = True, positive = True)
L = sp.symbols('L', real = True, positive = True)
K = sp.symbols('K', real = True, positive = True)
S = sp.symbols('S', real = True, positive = True)

#ZC = sp.symbols('ZC', real = True, positive = True)
#ZLC = sp.symbols('ZC//ZL', real = True, positive = True)

#Armo Tx1, Tx2, Tx3, Tg1 y Tg2 
Tx1 = sp.Matrix ([
    [1, 1/(S*C)],
    [0, 1]
    ])

Tx2 = sp.Matrix ([
    [1, (S/C)/(S*S+ (1/(L*C)) )],
    [0, 1]
    ])


Tx3 = Tx1

Tg1 = sp.Matrix ([
    [0, K],
    [1/K, 0]
    ])

Tg2 = Tg1

#Multiplico matrices
Tt = Tx1*Tg1*Tx2*Tg2*Tx3
#Reemplazo C y L por sus valores
Tt_fx = Tt.subs({C:0.5, L:2, K:np.sqrt(2)})
#Tt_fx = Tt.subs({ZC:1/(S*C), ZLC:(S/C)/(S*S+(1/(L*C)))})

Tt_fx

sp.simplify(Tt_fx)
```

Efectivamente, si en la transferencia obtenida analíticamente sumo los términos de cada elemento en una única fracción, llego a ese mismo resultado.
